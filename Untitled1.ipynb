```python

# ============================================================
# MOVIE RECOMMENDATION AND USER ANALYSIS - DWDM PROJECT
# Dataset: MovieLens 100K
# Visualizations + Collaborative Filtering Evaluation
# ============================================================

import os
import zipfile
import urllib.request
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.sparse import csr_matrix
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error

# ------------------------------------------------------------
# 1. DOWNLOAD AND LOAD MOVIELENS 100K
# ------------------------------------------------------------

url = "https://files.grouplens.org/datasets/movielens/ml-100k.zip"
zip_path = "/content/ml-100k.zip"

if not os.path.exists("/content/ml-100k"):
    urllib.request.urlretrieve(url, zip_path)
    with zipfile.ZipFile(zip_path, "r") as z:
        z.extractall("/content/")

data_path = "/content/ml-100k"

ratings = pd.read_csv(
    f"{data_path}/u.data",
    sep="\t",
    names=["user_id", "movie_id", "rating", "timestamp"]
)

movie_columns = [
    "movie_id", "title", "release_date", "video_release_date",
    "imdb_url"
] + [
    "genre_" + str(i) for i in range(19)
]

movies = pd.read_csv(
    f"{data_path}/u.item",
    sep="|",
    encoding="latin-1",
    names=movie_columns
)

genre_columns = movie_columns[5:]

print("Dataset loaded successfully!")
print("Ratings:", len(ratings))
print("Users:", ratings["user_id"].nunique())
print("Movies rated:", ratings["movie_id"].nunique())

# ------------------------------------------------------------
# 2. VISUALIZATION 1: RATING DISTRIBUTION
# ------------------------------------------------------------

rating_counts = (
    ratings["rating"]
    .value_counts()
    .reindex([1, 2, 3, 4, 5], fill_value=0)
)

plt.figure(figsize=(8, 5))
bars = plt.bar(
    rating_counts.index.astype(str),
    rating_counts.values
)

plt.title("Movie Rating Distribution")
plt.xlabel("Rating (1–5 Stars)")
plt.ylabel("Number of Ratings")
plt.bar_label(bars, fmt="%.0f")
plt.tight_layout()
plt.show()

print("Average rating:", round(ratings["rating"].mean(), 2))
print(
    "Percentage of 4- and 5-star ratings:",
    round(ratings["rating"].isin([4, 5]).mean() * 100, 2), "%"
)

# ------------------------------------------------------------
# 3. VISUALIZATION 2: AVERAGE RATING BY GENRE
# ------------------------------------------------------------

movie_genres = movies[["movie_id"] + genre_columns].melt(
    id_vars="movie_id",
    var_name="genre",
    value_name="included"
)

movie_genres = movie_genres[movie_genres["included"] == 1]
movie_genres["genre"] = movie_genres["genre"].str.replace(
    "genre_", "", regex=False
)

genre_ratings = ratings.merge(
    movie_genres[["movie_id", "genre"]],
    on="movie_id",
    how="inner"
)

genre_stats = (
    genre_ratings.groupby("genre")["rating"]
    .agg(["mean", "count"])
    .rename(columns={"mean": "average_rating", "count": "rating_count"})
)

# Show genres with at least 5,000 rating events
genre_stats = genre_stats[
    genre_stats["rating_count"] >= 5000
].sort_values("average_rating")

plt.figure(figsize=(9, 6))
bars = plt.barh(
    genre_stats.index,
    genre_stats["average_rating"]
)

plt.title("Average Rating by Genre (Minimum 5,000 Ratings)")
plt.xlabel("Average Rating (1–5)")
plt.ylabel("Genre")
plt.xlim(0, 5)
plt.bar_label(bars, fmt="%.2f", padding=3)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 4. VISUALIZATION 3: HIGHEST-RATED MOVIES
# Minimum 100 ratings per movie
# ------------------------------------------------------------

movie_stats = (
    ratings.groupby("movie_id")["rating"]
    .agg(["mean", "count"])
    .rename(columns={"mean": "average_rating", "count": "rating_count"})
    .reset_index()
)

movie_stats = movie_stats[movie_stats["rating_count"] >= 100]

movie_stats = movie_stats.merge(
    movies[["movie_id", "title"]],
    on="movie_id",
    how="left"
)

top_movies = movie_stats.sort_values(
    ["average_rating", "rating_count"],
    ascending=[False, False]
).head(10).sort_values("average_rating")

plt.figure(figsize=(10, 6))
bars = plt.barh(
    top_movies["title"],
    top_movies["average_rating"]
)

plt.title("Highest-Rated Movies (At Least 100 Ratings)")
plt.xlabel("Average Rating (1–5)")
plt.ylabel("Movie")
plt.xlim(0, 5)
plt.bar_label(bars, fmt="%.2f", padding=3)
plt.tight_layout()
plt.show()

print("\nTop-rated movies:")
print(top_movies[["title", "average_rating", "rating_count"]]
      .to_string(index=False))

# ------------------------------------------------------------
# 5. TRAIN/TEST SPLIT: 80% TRAIN, 20% TEST
# Random seed = 42
# ------------------------------------------------------------

train, test = train_test_split(
    ratings,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

print("\nTraining ratings:", len(train))
print("Testing ratings:", len(test))

# Use stable mappings from all dataset IDs.
# The model matrices are built from training ratings only.
user_ids = sorted(ratings["user_id"].unique())
movie_ids = sorted(ratings["movie_id"].unique())

user_to_idx = {u: i for i, u in enumerate(user_ids)}
movie_to_idx = {m: i for i, m in enumerate(movie_ids)}

n_users = len(user_ids)
n_movies = len(movie_ids)

train_users = train["user_id"].map(user_to_idx).to_numpy()
train_movies = train["movie_id"].map(movie_to_idx).to_numpy()
train_values = train["rating"].to_numpy(dtype=float)

# Matrix rows = movies, columns = users
R = csr_matrix(
    (train_values, (train_movies, train_users)),
    shape=(n_movies, n_users)
)

M = csr_matrix(
    (np.ones(len(train)), (train_movies, train_users)),
    shape=(n_movies, n_users)
)

# ------------------------------------------------------------
# 6. ITEM-BASED COLLABORATIVE FILTERING
# Cosine similarity using co-rated users only
# Require at least 10 shared raters
# ------------------------------------------------------------

print("\nCalculating item similarities...")

# Number of users who rated each pair of movies
co_raters = (M @ M.T).toarray()

# Dot product of ratings over shared users
dot_products = (R @ R.T).toarray()

# Sum of squared ratings restricted to users who rated
# both the target movie and the comparison movie.
R_squared = R.multiply(R)
shared_norms = (R_squared @ M.T).toarray()

denominator = np.sqrt(
    shared_norms * shared_norms.T
)

similarity = np.divide(
    dot_products,
    denominator,
    out=np.zeros_like(dot_products, dtype=float),
    where=denominator > 0
)

# Ignore self-similarity and pairs with fewer than 10 co-raters
similarity[co_raters < 10] = 0.0
np.fill_diagonal(similarity, 0.0)

# User's training-set ratings, used to predict held-out ratings
user_history = {
    user_to_idx[u]: list(zip(
        group["movie_id"].map(movie_to_idx).to_numpy(),
        group["rating"].to_numpy(dtype=float)
    ))
    for u, group in train.groupby("user_id")
}

global_mean = train["rating"].mean()

movie_means = train.groupby("movie_id")["rating"].mean().to_dict()

# ------------------------------------------------------------
# 7. PREDICT TEST RATINGS
# Use at least 2 valid neighbours; otherwise global mean
# ------------------------------------------------------------

print("Predicting held-out ratings...")

baseline_predictions = []
collaborative_predictions = []

for row in test.itertuples(index=False):
    u = user_to_idx[row.user_id]
    target = movie_to_idx[row.movie_id]
    actual_rating = float(row.rating)

    # Baseline: training movie mean; global mean for unseen movie
    baseline_predictions.append(
        movie_means.get(row.movie_id, global_mean)
    )

    history = user_history.get(u, [])

    if not history:
        collaborative_predictions.append(global_mean)
        continue

    rated_items = np.array([item for item, _ in history], dtype=int)
    rated_scores = np.array([score for _, score in history], dtype=float)

    sims = similarity[target, rated_items]

    # Only movies with valid similarity are neighbours
    valid = sims > 0
    sims = sims[valid]
    scores = rated_scores[valid]

    # Keep the strongest neighbours
    if len(sims) >= 2:
        order = np.argsort(sims)[::-1][:20]
        sims = sims[order]
        scores = scores[order]

        prediction = np.sum(sims * scores) / np.sum(np.abs(sims))
    else:
        prediction = global_mean

    collaborative_predictions.append(
        float(np.clip(prediction, 1, 5))
    )

# ------------------------------------------------------------
# 8. EVALUATION: MAE AND RMSE
# ------------------------------------------------------------

y_true = test["rating"].to_numpy(dtype=float)

baseline_mae = mean_absolute_error(y_true, baseline_predictions)
baseline_rmse = np.sqrt(mean_squared_error(y_true, baseline_predictions))

collab_mae = mean_absolute_error(y_true, collaborative_predictions)
collab_rmse = np.sqrt(mean_squared_error(y_true, collaborative_predictions))

results = pd.DataFrame({
    "Model": ["Movie-Mean Baseline", "Item-Based Collaborative Filtering"],
    "MAE": [baseline_mae, collab_mae],
    "RMSE": [baseline_rmse, collab_rmse]
})

print("\nMODEL EVALUATION RESULTS")
print(results.round(4).to_string(index=False))

# ------------------------------------------------------------
# 9. VISUALIZATION 4: MODEL COMPARISON
# Lower MAE/RMSE is better
# ------------------------------------------------------------

x = np.arange(len(results))
width = 0.35

fig, ax = plt.subplots(figsize=(9, 5))

bars1 = ax.bar(
    x - width / 2,
    results["MAE"],
    width,
    label="MAE"
)

bars2 = ax.bar(
    x + width / 2,
    results["RMSE"],
    width,
    label="RMSE"
)

ax.set_title("Model Evaluation: Movie Mean vs Collaborative Filtering")
ax.set_ylabel("Prediction Error (Rating Points)")
ax.set_xticks(x)
ax.set_xticklabels(
    ["Movie-Mean Baseline", "Item-Based CF"],
    rotation=10
)
ax.legend()
ax.bar_label(bars1, fmt="%.3f", padding=2)
ax.bar_label(bars2, fmt="%.3f", padding=2)

plt.tight_layout()
plt.show()

print("\nInterpretation:")
if collab_mae < baseline_mae and collab_rmse < baseline_rmse:
    print("Collaborative filtering has lower MAE and RMSE on this split.")
else:
    print("The movie-mean baseline is better on at least one error metric.")
    print("Do not claim collaborative filtering is better unless the results support it.")

print("\nAll visualizations and evaluation completed.")
```
